```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
# Crear conexión con mi db local (contenedor)

user = 'postgres'
password = 'postgres'
host = 'localhost'
port = '5432'
database = 'postgres'


db_url = f'postgresql+psycopg2://{user}:{password}@{host}:{port}/{database}'

engine = create_engine(db_url)

```

```python
# Carga del archivo

file_name = '../BC_A&A_with_ATD.csv'
df = pd.read_csv(file_name)

df.head()
```

```python
# Guardo el csv proporcionado como una tabla del ambiente dev

df.to_sql(
    name='mx_delivery_trips_info',
    schema='uber_dev',
    if_exists='append',
    index=False,
    con=engine
)
```
